```python
import pandas as pd
import numpy as np
```

```python
# Loading the data
df = pd.read_csv('customer_purchase_data.csv')
```

```python
# Short preview
df.head()
```

```python
# Basic Info 
df.info()
```

```python
# Applying descriptive stats for getting a general idea 
df.describe(include="all")
```

```python
# Checking for null values
df.isnull().sum()
```

```python
# Filling null values with median, but category-wise, because diff category have diff medians
df['Review_Rating'] = df.groupby('Category')['Review_Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Final Check
df.isnull().sum()
```

```python
# Converting column names into Snake-case, which make it more consistent,
# Easier to referencing and connecting with SQL and other tools.

df.columns = df.columns.str.lower()
df.columns
```

```python
# Creating a new column 'age_group'
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels) # qcut - splits the data in n parts
```

```python
# Preview
df[['age', 'age_group']].head(10)
```

```python
# Checking age_group distribution
df['age_group'].value_counts()
```

```python
# Creating a new column 'purchase_frequency_days', by mapping frequency_of_purchases with this dict
frequency_dict = {
    'Fortnightly': 14, 
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90, 
    'Bi-Weekly': 14, 
    'Annually': 365, 
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_dict)
```

```python
# Preview
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
# Checking for duplicate rows
print(f"Duplicate rows: {df.duplicated().sum()}")
df = df.drop_duplicates()
```

```python
# Checking for outliers in numeric columns
numeric_cols = ['age', 'purchase_amount', 'previous_purchases', 'review_rating']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
    print(f"{col}: {len(outliers)} outliers")
```
